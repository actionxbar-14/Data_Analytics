```python
# 5.  Array Reshaping and Flattening : 
```

```python
arr1 = np.array([1,2,3,4,5])
print(arr1)
```

```python
reshaped_arr1 = arr1.reshape((5, 1))  #-----> 5 rows  , 1 column
print(reshaped_arr1)
```

```python
reshaped2_arr1 = arr1.reshape((1, 5))  #-----> 1 rows  , 5 column
print(reshaped_arr1)
```

```python
## 6. Array Stacking and Splitting : 
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.vstack((a,b)))   # --> vertical stacking : row wise
print(np.hstack((a,b)))   # --> Horizontal stacking : column wise
```

```python
# :--> Spiliting : 
```

```python
c = np.array([ [1,2,3] , [4,5,6] ])
print(c)
```

```python
hsplit = np.hsplit(c , 3)   #--> horizontal split

for s in hsplit:
    print(s)
```

```python
vsplit = np.vsplit(c, 2)  #--> vertical split 

for s in vsplit:
    print(s)
```

```python
# Ex :
```

```python
a = np.array([1,2,3,4])
b = np.array([5])
print(a + b)
```
