```python
## 1. Creating arrays : 
```

```python
list1 = [1,2, 3, 4]
print(list1)
```

```python
arr_list1 = np.array(list1)
print(arr_list1)
```

```python
print(type(arr_list1))
print(arr_list1.ndim)
```

```python
list1 = [1,2,4]
list2 = [4,5,6]
arr_list1_2 = np.array([list1, list2])
print(arr_list1_2)
print(arr_list1_2.ndim)
```

```python
list1 = [1,2,3]
list2 = [4,5,6]
list3 = [7,8,9]
arr_list1_3 = np.array([[list1 , list2 , list3]])
print(arr_list1_3)
print(arr_list1_3.ndim)
```

```python
## 3. Array initialization methods : 
```

```python
#--> Zeros array :
```

```python
zero_arr = np.zeros((2,3))
print(zero_arr)
```

```python
#--> Ones array :
```

```python
ones_arr = np.ones((1 , 4))
print(ones_arr)
```

```python
#--> Full array : 
```

```python
full_arr = np.full((3,2),7)
print(full_arr)
```

```python
# Identity Matrix : 3 rows , 3 columns
```

```python
id_arr = np.eye(3)
print(id_arr)
```

```python
# Empty array : 
```

```python
print(np.empty(2))
```

```python
# evenly spaced array : 
```

```python
print(np.arange(1, 10 ,2))
```

```python
# specific number of equally (spaced) values between a range : 
```

```python
print(np.linspace(1 , 10 , 4))
```

```python
# random values array :   [float]
```

```python
r_arr = np.random.rand(3,2)
print(r_arr)
```

```python
# random values array : [int]
```

```python
rint_arr  = np.random.randint(1 , 100 , (3 ,3))
print(rint_arr)
```

```python
## 4. array Indexing and Slicing : 
```

```python
a = np.array([1,3,5,7,9])
print(a)
```

```python
print(a[0])
```

```python
print(a[-1])
```

```python
#-->Slicing : 
#->array[start:stop:step]
#->stop - index , that is not included in output
```

```python
a = np.array([1,3,5,7,9])
print(a)
```

```python
print(a[-3:])
```
