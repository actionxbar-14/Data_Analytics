```python
import pandas as pd
import random
```

```python
## --> One Dice thrown : 

# L = []
# for i in range(10000):
#     L.append(random.randint(1,6))


## --> Two Dice thrown : 

L = []
for i in range(10000):
    a = random.randint(1 , 6)
    b = random.randint(1 , 6)

    L.append(a + b)
    
```

```python
len(L)
```

```python
L[:5]
```

```python
s = (pd.Series(L).value_counts() / pd.Series(L).value_counts().sum()).sort_index()
s
```

```python
s.plot(kind = 'bar')

```

```python
import numpy as np

np.cumsum(s)
```

```python
## Commulative Mass Function ( CMF ) :

s.plot(kind='bar')
```

```python
## cummulative distribution function ( CDF ) :

np.cumsum(s).plot(kind='bar')
```

##  Parametric Density Estimations : 

```python
import matplotlib.pyplot as plt
import numpy as np
from numpy.random import normal

sample = normal(loc=50, scale=5,size=1000)
```

```python
sample
```

```python
sample.mean()
```

```python
# plot histogram to understand the distribution of data :

plt.hist(sample , bins = 10)
```

```python
# calculate sample mean and sample std dev :

sample_mean = sample.mean()
sample_std = sample.std()
```

```python
# fit the distribution with the above parameters : 

from scipy.stats import norm
dist = norm(sample_mean, sample_std)


```

```python
values = np.linspace(sample.min(),sample.max(),100)
```

```python
sample.max()
```

```python
probabilities = [dist.pdf(value) for value in values]
```

```python
# plot the histogram and pdf :


plt.hist(sample,bins=10,density=True)
plt.plot(values,probabilities)
```

```python
import seaborn as sns
sns.distplot(sample)
```

## Non - Parametric Density Estimations : 

```python
### KDE : 
```

```python
# generate a sample
sample1 = normal(loc=20, scale=5, size=300)
sample2 = normal(loc=40, scale=5, size=700)
sample = np.hstack((sample1, sample2))
```

```python
sample
```

```python
# plot histogram bins=50
plt.hist(sample,bins=50)
```

```python
from sklearn.neighbors import KernelDensity

model = KernelDensity(bandwidth=5, kernel='gaussian')

# convert data to a 2D array
sample = sample.reshape((len(sample), 1))
```

```python

model.fit(sample)
```

```python
values = np.linspace(sample.min(),sample.max(),100)
values = values.reshape((len(values), 1))
```

```python
probabilities = model.score_samples(values)
probabilities = np.exp(probabilities)



'''
score_samples(values) returns the log-density estimate of the input samples values.
This is because the score_samples() method of the KernelDensity class returns the logarithm of 
the probability density estimate rather than the actual probability density estimate.


'''


```

```python
plt.hist(sample, bins=50, density=True)
plt.plot(values[:], probabilities)
plt.show()
```

```python
import seaborn as sns

sns.kdeplot(sample.reshape(1000),bw_adjust=0.02)
```
