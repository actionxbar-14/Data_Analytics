## Annotations : 

```python
## graph ke upperr ek text ko place krta hai.
```

```python
batters = pd.read_csv(r'C:\Users\ANUBHAV\Data_Analytics\Python\Matplotlib\Datasets\datasets-session-24\batter.csv')
batters.head()
```

```python
batters.shape
```

```python
sample_df = batters.head(100).sample(25 , random_state = 5)

sample_df
```

```python
plt.figure(figsize = (18 , 8))

plt.scatter(sample_df['avg'] , sample_df['strike_rate'] , s = sample_df['runs'])

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.scatter(x , y)

plt.text(1 , 5  , 'Point 1')
plt.text(2 , 6  , 'Point 2')
plt.text(3 , 7  , 'Point 3')
plt.text(4 , 8  , 'Point 4' , fontdict = {'size' : 12 , 'color' : 'brown'})
```

## Horizontal and Vertical lines : 

```python
plt.figure(figsize = (18 , 8))

plt.scatter(sample_df['avg'] , sample_df['strike_rate'] , s = sample_df['runs'])

plt.axhline(130 , color = 'red')  #--> seggreate data through horizontal line
plt.axhline(140 , color = 'green')  #--> seggreate data through horizontal line
plt.axvline(30 , color = 'red')   #--> seggreate data through vertical line




for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
## different way to plot :

fig , ax = plt.subplots()

ax.scatter(batters['avg'] , batters['strike_rate'] , color = 'red')
ax.set_title('Something')
ax.set_xlabel('Avg')
ax.set_ylabel('Strike Rate')

fig.show()
```

```python
fig , ax = plt.subplots(nrows = 2 , ncols = 1 , sharex = True , figsize = (12,8))



ax[0].scatter(batters['avg'] , batters['strike_rate'] , color = 'red')
ax[1].scatter(batters['avg'] , batters['runs'])


ax[0].set_title('Avg vs strike rate')
ax[0].set_ylabel('Strike rate')


ax[1].set_title('Avg vs Runs')
ax[1].set_ylabel('Runs')
ax[0].set_xlabel('Avg')
```

## 3D Scatter Plots : 

```python
batters


fig = plt.figure()

ax = plt.subplot(projection = '3d')
ax.scatter3D(batters['runs'] , batters['avg'] , batters['strike_rate'])
ax.set_title('IPL batsman analysis')

ax.set_xlabel('Runs')
ax.set_ylabel('Avg')
ax.set_zlabel('SR')
```

## 3D Line Plot : 

```python
x = [0 , 1 , 5 , 25]
y = [0 , 10 , 13 , 0 ]
z = [0 , 13 , 20 , 9]
fig = plt.figure()

ax = plt.subplot(projection = '3d')
ax.scatter3D(x, y , z , s = [100 , 100, 100 , 100])
ax.plot3D(x ,y , z , color = 'red')
```
