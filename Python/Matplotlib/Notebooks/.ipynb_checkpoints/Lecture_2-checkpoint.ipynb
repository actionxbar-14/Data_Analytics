```python
batters = pd.read_csv(r'C:\Users\ANUBHAV\Data_Analytics\Python\Matplotlib\Datasets\datasets-session-24\batter.csv')
batters.head()
```

```python
sample_df = batters.head(100).sample(25 , random_state = 5)

sample_df
```

```python
plt.figure(figsize = (18 , 8))

plt.scatter(sample_df['avg'] , sample_df['strike_rate'] , s = sample_df['runs'])

plt.axhline(130 , color = 'red')  #--> seggreate data through horizontal line
plt.axhline(140 , color = 'green')  #--> seggreate data through horizontal line
plt.axvline(30 , color = 'red')   #--> seggreate data through vertical line




for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

## 3D Surface Plots : 

```python
x = np.linspace(-10 , 10 , 100)
y = np.linspace(-10 , 10 , 100)
```

```python
xx , yy = np.meshgrid(x,y)
# xx.shape   #--> mesh grid
yy.shape #--> mesh grid
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize = (12 , 8))

ax = plt.subplot(projection = '3d')
p = ax.plot_surface(xx , yy , z , cmap = 'viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize = (12 , 8))

ax = plt.subplot(projection = '3d')
p = ax.plot_surface(xx , yy , z , cmap = 'viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.log(yy)

fig = plt.figure(figsize = (12 , 8))

ax = plt.subplot(projection = '3d')
p = ax.plot_surface(xx , yy , z , cmap = 'viridis')
fig.colorbar(p)
```

## Contour Plots : 

```python
## 3d graph ko 2d mai represent krta hai. 
```

```python
z = xx**2 + yy**2
z.shape


fig = plt.figure(figsize = (12 , 8))

ax = plt.subplot(projection = '3d')
p = ax.plot_surface(xx , yy , z , cmap = 'viridis')
fig.colorbar(p)
```

```python
z = xx**2 + yy**2
z.shape


fig = plt.figure(figsize = (12 , 8))

ax = plt.subplot(projection = '3d')
p = ax.contour(xx , yy , z , cmap = 'viridis')
fig.colorbar(p)
```

```python
z = xx**2 + yy**2
z.shape


fig = plt.figure(figsize = (12 , 8))

ax = plt.subplot(projection = '3d')
p = ax.contourf(xx , yy , z , cmap = 'viridis')
fig.colorbar(p)
```

```python
z = xx**2 + yy**2
z.shape


fig = plt.figure(figsize = (12 , 8))

ax = plt.subplot()
p = ax.contourf(xx , yy , z , cmap = 'viridis')
fig.colorbar(p)
```
