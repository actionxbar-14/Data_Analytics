## plotting multiple plots :

## labels title : 

```python


plt.plot(batsman['index'] , batsman['V Kohli'])
plt.plot(batsman['index'] , batsman['RG Sharma'])

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

```

##  colors(hex)  : 

```python



# plt.plot(batsman['index'],batsman['V Kohli'],color='green')
# plt.plot(batsman['index'],batsman['RG Sharma'],color='black')

plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

## line(width and style) :  

```python


plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F' ,  linestyle =  'dashed')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6', linestyle =  'dotted')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

## marker(size) : 

```python



# plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F' ,  linestyle =  'solid' , linewidth = 4)
# plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6', linestyle =  'dotted' ,  linewidth = 2)


plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F' ,  linestyle =  'solid' , linewidth = 4 , marker = 'D' , markersize = 5) # --> [ D - Diamond ]
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6', linestyle =  'dotted' ,  linewidth = 2 ,  marker = '.' , markersize = 15)


plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')


```

## legend ( location ) : 

```python




plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F' ,  linestyle =  'solid' , linewidth = 4 , marker = 'D' , markersize = 5 , label = 'virat') # --> [ D - Diamond ]
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6', linestyle =  'dotted' ,  linewidth = 2 ,  marker = '.' , markersize = 15 , label = 'kohli')


plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')


#--> legend function to print label  :

plt.legend()

plt.legend(loc = 'upper right') 

```

## limiting axes : 

```python


price = [48000 , 54000 , 57000 , 49000 , 47000 , 4500000]   #-->  [ 4500000 : outlier ] 
year  = [2015 , 2016 , 2017 , 2018 , 2019 , 2020]  

plt.plot(year , price)

plt.ylim(0 , 75000)
plt.xlim(2017 , 2019)
```

## grid : 

```python


price = [48000 , 54000 , 57000 , 49000 , 47000 , 4500000]   #-->  [ 4500000 : outlier ] 
year  = [2015 , 2016 , 2017 , 2018 , 2019 , 2020]  

plt.plot(year , price)

plt.grid()
```

## show : 

```python
#--> plt.show is the main function jiske wajah se actul coding enviornement mai graph plot hote hai.
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F' ,  linestyle =  'solid' , linewidth = 4 , marker = 'D' , markersize = 5) # --> [ D - Diamond ]
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6', linestyle =  'dotted' ,  linewidth = 2 ,  marker = '.' , markersize = 15)


plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.grid()

plt.show()
```

## plt.scatter on pandas data :

### size : 

```python
tips = sns.load_dataset('tips')
tips

plt.scatter(tips['total_bill'], tips['tip'])


#--> s parameter :



plt.scatter(tips['total_bill'], tips['tip'] , s = tips['size'] * 20)  #--> This is a slower technique
```

###  scattterplot using plt.plot : 


```python



plt.plot(tips['total_bill'] , tips['tip'])

plt.plot(tips['total_bill'] , tips['tip'] , 'o')  #--> This is a faster technique
```

## Bar chart : 




image.png

```python
# - Bivariate Analysis.
# - Numerical vs Categorical.
# - Use case - Aggregate analysis of groups.
```

```python
## simple bar chart  :

children = [10 ,20 ,30 ,25 ,20]
colors = ['red' , 'blue' , 'green' , 'yellow' , 'pink']


plt.bar(colors , children , color = 'red')
```

```python
## horizontal bar chart : 

plt.barh(colors,children,color='red')
```

```python
## color and label : 

df = pd.read_csv(r'C:\Users\ANUBHAV\Data_Analytics\Python\Matplotlib\Datasets\datasets-session-23\batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'] , df['2015'])
plt.bar(df['batsman'] , df['2016'])


```

```python
plt.bar(np.arange(df.shape[0]) - 0.2, df['2015'] , width = 0.2 , color = 'yellow')
plt.bar(np.arange(df.shape[0]), df['2016'] , width = 0.2 , color = 'blue')
plt.bar(np.arange(df.shape[0]) + 0.2 , df['2017'] , width = 0.2 , color = 'red')


plt.xticks(np.arange(df.shape[0]), df['batsman'])

plt.show()
```

```python
## xticks problem : 

# a problem
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children,color='black')
plt.xticks(rotation='vertical')

```

```python
## Stacked bar chart : 

plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015')

plt.legend()
plt.show()
```

## Pie chart



image.png

```python
## --> categories ke upper plot kiya jata hai , or ye batata hai ki kisi particular cateogory ka 100% ya total percentage
##      mai se kitna percent contribution hai.
```

```python
# - Univariate/Bivariate Analysis.
# - Categorical vs numerical.
# - Use case - To find contibution on a standard scale.
```

```python
## simple data :

data = [24 , 45 , 100 , 20 , 49]
subjects = ['eng' , 'science' , 'maths' , 'sst' , 'hindi']
plt.pie(data ,  labels = subjects)
plt.show()
```

```python
## with dataset : 

df = pd.read_csv(r'C:\Users\ANUBHAV\Data_Analytics\Python\Matplotlib\Datasets\datasets-session-23\gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'] , labels = df['batsman'] , autopct = '%0.1f%%')
plt.show()
```

```python
## percentage and colors : 

plt.pie(df['batsman_runs'] , labels = df['batsman'] , autopct = '%0.1f%%' , colors = ['blue' , 'green' , 'yellow' , 'pink' , 'red'])
plt.show()
```

```python

##  explode shadow
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.3,0,0,0,0,0.1],shadow=True)
plt.show()
```
