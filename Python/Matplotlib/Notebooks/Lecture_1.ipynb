## plotting multiple plots : 

```python


plt.plot(batsman['index'] , batsman['V Kohli'])
plt.plot(batsman['index'] , batsman['RG Sharma'])
```

## labels title : 

```python


plt.plot(batsman['index'] , batsman['V Kohli'])
plt.plot(batsman['index'] , batsman['RG Sharma'])

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

```

##  colors(hex)  : 

```python



# plt.plot(batsman['index'],batsman['V Kohli'],color='green')
# plt.plot(batsman['index'],batsman['RG Sharma'],color='black')

plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

## line(width and style) :  

```python


plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F' ,  linestyle =  'dashed')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6', linestyle =  'dotted')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

## marker(size) : 

```python



# plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F' ,  linestyle =  'solid' , linewidth = 4)
# plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6', linestyle =  'dotted' ,  linewidth = 2)


plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F' ,  linestyle =  'solid' , linewidth = 4 , marker = 'D' , markersize = 5) # --> [ D - Diamond ]
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6', linestyle =  'dotted' ,  linewidth = 2 ,  marker = '.' , markersize = 15)


plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')


```

## legend ( location ) : 

```python




plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F' ,  linestyle =  'solid' , linewidth = 4 , marker = 'D' , markersize = 5 , label = 'virat') # --> [ D - Diamond ]
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6', linestyle =  'dotted' ,  linewidth = 2 ,  marker = '.' , markersize = 15 , label = 'kohli')


plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')


#--> legend function to print label  :

plt.legend()

plt.legend(loc = 'upper right') 

```

## limiting axes : 

```python


price = [48000 , 54000 , 57000 , 49000 , 47000 , 4500000]   #-->  [ 4500000 : outlier ] 
year  = [2015 , 2016 , 2017 , 2018 , 2019 , 2020]  

plt.plot(year , price)

plt.ylim(0 , 75000)
plt.xlim(2017 , 2019)
```
