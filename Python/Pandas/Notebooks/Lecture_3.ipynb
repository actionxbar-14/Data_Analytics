# GroupBy Introduction : 



```python
# Group By : 
#### --> Group by mai ham data ke ander groups form karte hai , on the basis of some column.
#### --> Group by hamesha categorical columns prr hi lgta hai. 
```

```python
movies.head()
```

```python
genres = movies.groupby('Genre')
genres
```

```python
## Applying built-in aggregation function on groupby objects :

genres.sum()
genres.min()
genres.max()
```

```python
## Find the top 3 genres by total earning : 

movies.groupby('Genre').sum()['Gross'].sort_values(ascending=False).head(3)
```

```python
movies.groupby('Genre')['Gross'].sum().sort_values(ascending=False).head(3)
```

```python
## Find the genre with highest avg IMDB rating : 

movies.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False).head(1)


```

```python
## Find director with most popularity : 

movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
## Find the highest rated movie of each genre : 

movies.groupby('Genre')['IMDB_Rating'].max()
```

```python
## Find number of movies done by each actor :

movies.groupby('Star1')['Series_Title'].count().sort_values(ascending=False)

```

# GroupBy Attributes And Methods : 


## find total number of groups -> len :

```python
len(movies.groupby('Genre'))


# alternative :
movies['Genre'].nunique()
```

## find the total number of rows in each group -> size :


```python
movies.groupby('Genre').size()


# alternative :
movies['Genre'].value_counts()
```

## first() / last() / nth : 

```python
genres = movies.groupby('Genre')
```

```python
genres.first() #--> har ak genre ki first movie ko nikl ke dega.

genres.last() #--> har ak genre ki last movie ko nikl ke dega.

genres.nth(5) #--> har ek genre ki 5th movie nikal krr dega.

```

## get_group  : 

```python
#--> get_group kisi particular 'Genre' ke bare mai sari details deta hai.

genres.get_group('Horror')
genres.get_group('Fantasy')


#--> Alternative: 
movies[movies['Genre'] == 'Horror']
movies[movies['Genre'] == 'Fantasy']
```

```python
# groups attribute : 
#--> har ak genre ki value ko as a index store krta hai.

genres.groups
```

## aggregate method : 


```python
# ak sath multiple aggregate method lga deta hai.
```

```python
# passing dict : 

genres.agg(
    {
        'Runtime' : 'mean',
        'IMDB_Rating' : 'mean',
        'No_of_Votes'  : 'sum',
        'Gross'  : 'sum',
        'Metascore' : 'min'
    }
)
```

```python
# passing list :

genres.agg(['min' , 'max' , 'sum'])
```

```python
# merging passing dict + passing list result : 

genres.agg(
    {
        'Runtime' : ['mean' , 'max'],
        'IMDB_Rating' : 'mean',
        'No_of_Votes'  : ['sum',  'max'],
        'Gross'  : 'sum',
        'Metascore' : 'min'
    }
)
```
