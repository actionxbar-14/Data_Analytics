# Filtering a DataFrame : 

```python
ipl.head(2)

```

```python
##--> find all the final winners  :


mask = ipl['MatchNumber'] == 'Final'

new_df = ipl[mask]

new_df[ ['Season' , 'WinningTeam'] ]
```

```python
ipl[ipl['MatchNumber'] == 'Final'][ ['Season' , 'WinningTeam'] ]
```

```python
## How many super over finishes have occured : 

ipl[ipl['SuperOver'] == 'Y']


ipl[ipl['SuperOver'] == 'Y'].shape[0]
```

```python
## How many matches has csk won in kolkata : 

ipl[ (ipl['City'] == 'Kolkata') &  (ipl['WinningTeam'] == 'Chennai Super Kings') ].shape[0]

```

```python
## toss winner is match winner in percentage  :

ipl[ ipl['TossWinner'] == ipl['WinningTeam'] ][ ['TossWinner' , 'WinningTeam'] ].shape[0]


ipl[ ipl['TossWinner'] == ipl['WinningTeam'] ][ ['TossWinner' , 'WinningTeam'] ].shape[0] / ipl.shape[0] * 100
```

```python
## movies with rating higher than 8 and votes > 10000 : 

movies[ (movies['imdb_rating'] > 8)  &  (movies['imdb_votes'] > 10000) ][ ['title_x' , 'imdb_rating' , 'imdb_votes'] ].shape[0]


# & ('imdb_votes' > 10000) ] 
```

```python
## Action movies with rating higher than 7.5 : 


# method_1 : 
movies[ (movies['imdb_rating'] > 7.5) &  (movies['genres'].str.split('|').apply(lambda x:'Action' in x)) ][ [ 'title_x' ,'genres' , 'imdb_rating'] ]

                                      
# method_2 : 
movies[ (movies['imdb_rating'] > 7.5) &  (movies['genres'].str.contains('Action')) ][ [ 'title_x' ,'genres' , 'imdb_rating'] ].shape[0]

```
