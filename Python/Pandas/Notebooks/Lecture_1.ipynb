## Series from lists : 

```python
# String : 

country = [ 'India',  'Pakistan' , 'USA' , 'Nepal' , 'SriLanka']

pd.Series(country)
```

```python
# integers : 

runs = [14 , 24 ,56 , 78 , 100]

pd.Series(runs)
```

```python
# custom index : 

marks = [67 , 57 , 89 , 100]
subjects = ['maths' , 'english' , 'science' , 'hindi']

pd.Series(marks)  #--> index default bnega.

pd.Series(marks , index = subjects)  #--> index as a subject bnega
```

```python
runs = pd.Series(marks)  
```

```python
# setting a name : 

marks = pd.Series(marks , index = subjects , name = 'anubhav ke marks')
marks
```

### drop_duplicates : 

```python
# ye phle wale value ko chodhkrr baki same value ke subsequent duplicate values ko drop krr deta hai. 
# but agar hame last occurence wale ko rkhna hai and baki saare occurence ko delete krna hai toh 'keep' ka use krna hota h.
```

```python
temp = pd.Series([1,1,2,2,3,3,4,4])
temp
```

```python
# drop_duplicates [ kepping first occurence ] : 
temp.drop_duplicates()
```

```python
# drop_duplicates[ keeping last occurence ] :

temp.drop_duplicates(keep = 'last')
```

```python
# --> check for duplicated values :  [ duplicated() ] :

temp.duplicated()

print(temp.duplicated().sum())
```
