```python
## --> Problem in vectorized operations in vanilla python : 

s = ['cat' , 'mat' , None , 'rat']

[i.startswith('c') for i in s]


#--> 'NoneType' object has no attribute 'startswith'.
#--> Relatively Slow.
```

```python
## How pandas solves this issue? 

s = pd.Series(['cat' , 'mat' , None, 'rat'])
s

# string accessor :
s.str.startswith('c')    #--> fast and optimized!
```

## Common Functions : 

### lower / upper / capitalize / title : 

```python

df['Name'].str.upper()
df['Name'].str.capitalize()
df['Name'].str.lower()
df['Name'].str.title()
```

### len / strip : 

```python
df['Name'][df['Name'].str.len()]
df['Name'][df['Name'].str.len() == 82].values[0]


"                             nitish                    ".strip()
df['Name'].str.strip()
```

### split : 

```python
df['Name'].str.split(',')
```

```python
df['lastname'] = df['Name'].str.split(',').str.get(0)  #--> Name
df.head()
```

```python
df[ ['title' , 'firstname'] ] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ' , n = 1 , expand = True)  
#--> [ ( n=1 ) 1st space prr hi split hoga , ( expand = True ) dataframe conversion]

df.head()
```

```python
df['title'].value_counts()
```

### replace : 

```python
df['title'] = df['title'].str.replace('Ms.','Miss.')
df['title'] = df['title'].str.replace('Mlle.','Miss.')
```

```python
df['title'].value_counts()
```

### startswith / endswith / isdigit / isalpha :

```python
## startswith / endswith :

df[df['firstname'].str.startswith('A')]

df[df['firstname'].str.endswith('A')]
```

```python
## isdigit / isalpha : 
```

```python
df[df['firstname'].str.isdigit()]
```

### Applying regex / contains :

```python
## search john -> both case : 

df[df['firstname'].str.contains('john' , case = False)]
```

```python
## find lastnames with start and end character is vowel : 

df[df['lastname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]

## find lastnames with start and end character is consonant : 

df[df['lastname'].str.contains('^[^[aeiouAEIOU].+[aeiouAEIOU]$')]
```
