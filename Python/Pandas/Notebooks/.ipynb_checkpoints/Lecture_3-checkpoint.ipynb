# GroupBy Introduction : 



```python
# Group By : 
#### --> Group by mai ham data ke ander groups form karte hai , on the basis of some column.
#### --> Group by hamesha categorical columns prr hi lgta hai. 
```

```python
movies.head()
```

```python
genres = movies.groupby('Genre')
genres
```

```python
## Applying built-in aggregation function on groupby objects :

genres.sum()
genres.min()
genres.max()
```

```python
## Find the top 3 genres by total earning : 

movies.groupby('Genre').sum()['Gross'].sort_values(ascending=False).head(3)
```

```python
movies.groupby('Genre')['Gross'].sum().sort_values(ascending=False).head(3)
```

```python
## Find the genre with highest avg IMDB rating : 

movies.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False).head(1)


```

```python
## Find director with most popularity : 

movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
## Find the highest rated movie of each genre : 

movies.groupby('Genre')['IMDB_Rating'].max()
```

```python
## Find number of movies done by each actor :

movies.groupby('Star1')['Series_Title'].count().sort_values(ascending=False)

```

# GroupBy Attributes And Methods : 


## find total number of groups -> len :

```python
len(movies.groupby('Genre'))


# alternative :
movies['Genre'].nunique()
```

## find the total number of rows in each group -> size :


```python
movies.groupby('Genre').size()


# alternative :
movies['Genre'].value_counts()
```

## first() / last() / nth : 

```python
genres = movies.groupby('Genre')
```

```python
genres.first() #--> har ak genre ki first movie ko nikl ke dega.

genres.last() #--> har ak genre ki last movie ko nikl ke dega.

genres.nth(5) #--> har ek genre ki 5th movie nikal krr dega.

```

## aggregate method : 


```python
# ak sath multiple aggregate method lga deta hai.
```

```python
# passing dict : 

genres.agg(
    {
        'Runtime' : 'mean',
        'IMDB_Rating' : 'mean',
        'No_of_Votes'  : 'sum',
        'Gross'  : 'sum',
        'Metascore' : 'min'
    }
)
```

```python
# passing list :

genres.agg(['min' , 'max' , 'sum'])
```

```python
# merging passing dict + passing list result : 

genres.agg(
    {
        'Runtime' : ['mean' , 'max'],
        'IMDB_Rating' : 'mean',
        'No_of_Votes'  : ['sum',  'max'],
        'Gross'  : 'sum',
        'Metascore' : 'min'
    }
)
```

## looping on groups : 

```python
genres
```

```python
for group , data in genres:
    print(data)
    # print(type(group) , type(data))
```

```python
## find the highest rated movie for each genres : 
df_list = []

for group, data in genres:
    df_list.append(
        data[data['IMDB_Rating'] == data['IMDB_Rating'].max()]
    )

df = pd.concat(df_list, ignore_index=True)

df
```

## split ( apply ) combine : 

```python
genres.apply(min)
```

```python
## find number of movies starting with 'A' for each group : 

def foo(group):
    return group['Series_Title'].str.startswith('A').sum()
```

```python
genres.apply(foo)
```

```python
## find ranking of each movie in the group according to the IMDB score :

def rank_movie(group):
    group['genre_rank'] = group['IMDB_Rating'].rank(ascending=False)
    return group
    
```

```python
genres.apply(rank_movie)
```

```python
## find normalized IMDB rating group wise : 


def normal(group):
    group['norm_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min()) /  ( group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
    return group
genres.apply(normal)
```

```python
ipl = pd.read_csv(r'C:\Users\ANUBHAV\Data_Analytics\Python\Pandas\Datasets\dataset-session-19\deliveries.csv')
ipl.head()
```

```python
ipl.shape
```

```python
## Find the top 10 batsman in terms of runs : 

ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)

```

```python
## find the batsman name with maximum number of sixes : 

six = ipl[ipl['batsman_runs'] == 6]
six.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1).index[0]
```

```python
## find batsman name with most number of 4's and 6's in last 5 overs :

temp_df = ipl[ipl['over'] > 15]
temp_df[ (temp_df['batsman_runs'] == 4) | (temp_df['batsman_runs'] == 6) ] 

temp_df.groupby('batsman')['batsman'].count().sort_values(ascending=False).head().index[0]

```

```python
## find V kohli's record against all teams : 

temp_df = ipl[ipl['batsman'] == 'V Kohli']

virat_stats = temp_df.groupby('bowling_team')['batsman_runs'].sum().reset_index()

virat_stats
```

```python
## Create a function that can return the highest score of any batsman : 

```

```python
def highest(batsman):
    temp_df = ipl[ipl['batsman'] == batsman]
    return temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
    
```

```python
highest('DA Warner')
```
