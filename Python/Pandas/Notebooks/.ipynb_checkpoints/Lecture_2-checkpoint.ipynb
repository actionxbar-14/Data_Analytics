# Creating DataFrame : 

## using 2D lists : 

```python
student_data = [
    [100 , 80 , 10] ,
    [90 ,  70 , 7] , 
    [120 , 100 , 14],
    [80 , 50 , 2]
]

pd.DataFrame(student_data , columns = ['iq' , 'marks' , 'package'] ) 
```

## using dicts  : 

```python
student_dict = {
    'name' : ['nitish' , 'ankit' , 'rupesh' , 'rishabh' , 'amit' , 'ankita'],
    'iq'  : [100 , 90 , 120 , 80 ,0 , 0],
    'marks' : [80 , 70 , 100 , 50, 0 , 0],
    'package' : [10 , 7 , 14 , 2 ,0 ,0]
}

students = pd.DataFrame(student_dict)
students.set_index('name' , inplace = True)  #--> kisi ak particular col ko as a index banata hai dataframe mai
students
```

# Selecting rows from a DataFrame : 



```python
# --> **iloc**  :  searches using index positions.
# --> **loc**   :  searches using index labels.
```

## iloc : 

```python
## last given value is exclusive of the provided range.
```

### Single Row : 


```python
movies.iloc[0] #--> index position ke hisab se data fetch krke dega
```

```python
type(movies.iloc[0])
```

```python
# --> NOTE : 
# 1. Single row nikalenge toh series milegi,
# 2. Multiple rows nikalenge toh dataFrame milega.
```

### Multiple Row : 

```python
movies.iloc[0:5]
```

### fancy indexing : 

```python


movies.iloc[ [0,4,8 ,16] ]
```

## loc : 

```python
## last given value is inclusive of the provided range.
```

```python
students
```

### Single items : 

```python
students.loc['nitish']
```

### Multiple items : 

```python
students.loc['nitish' : 'rishabh']
```

```python
students.loc['nitish' : 'rishabh' : 2]
```

### fancy indexing : 

```python
students.loc[['nitish' , 'ankita' , 'rupesh']]
```
