# Filtering a DataFrame : 

```python
ipl.head(2)

```

```python
##--> find all the final winners  :


mask = ipl['MatchNumber'] == 'Final'

new_df = ipl[mask]

new_df[ ['Season' , 'WinningTeam'] ]
```

```python
ipl[ipl['MatchNumber'] == 'Final'][ ['Season' , 'WinningTeam'] ]
```

```python
## How many super over finishes have occured : 

ipl[ipl['SuperOver'] == 'Y']


ipl[ipl['SuperOver'] == 'Y'].shape[0]
```

```python
## How many matches has csk won in kolkata : 

ipl[ (ipl['City'] == 'Kolkata') &  (ipl['WinningTeam'] == 'Chennai Super Kings') ].shape[0]

```

```python
## toss winner is match winner in percentage  :

ipl[ ipl['TossWinner'] == ipl['WinningTeam'] ][ ['TossWinner' , 'WinningTeam'] ].shape[0]


ipl[ ipl['TossWinner'] == ipl['WinningTeam'] ][ ['TossWinner' , 'WinningTeam'] ].shape[0] / ipl.shape[0] * 100
```

```python
## movies with rating higher than 8 and votes > 10000 : 

movies[ (movies['imdb_rating'] > 8)  &  (movies['imdb_votes'] > 10000) ][ ['title_x' , 'imdb_rating' , 'imdb_votes'] ].shape[0]


# & ('imdb_votes' > 10000) ] 
```

```python
## Action movies with rating higher than 7.5 : 


# method_1 : 
movies[ (movies['imdb_rating'] > 7.5) &  (movies['genres'].str.split('|').apply(lambda x:'Action' in x)) ][ [ 'title_x' ,'genres' , 'imdb_rating'] ]

                                      
# method_2 : 
movies[ (movies['imdb_rating'] > 7.5) &  (movies['genres'].str.contains('Action')) ][ [ 'title_x' ,'genres' , 'imdb_rating'] ].shape[0]

```

## sort_values : 

```python
# both series and dataFrame. 
# ascending -> na_position -> inplace -> multiple cols. 
```

```python
x = pd.Series([12,14,1,56,89])
x
```

```python
x.sort_values(ascending = False) #--> Series


```

```python
movies.head(4)
```

```python
movies.sort_values('title_x' , ascending=False)    #-->  DataFrame 
```

```python
students = pd.DataFrame(
    {
        'name':['nitish','ankit','rupesh',np.nan,'mrityunjay',np.nan,'rishabh',np.nan,'aditya',np.nan],
        'college':['bit','iit','vit',np.nan,np.nan,'vlsi','ssit',np.nan,np.nan,'git'],
        'branch':['eee','it','cse',np.nan,'me','ce','civ','cse','bio',np.nan],
        'cgpa':[6.66,8.25,6.41,np.nan,5.6,9.0,7.4,10,7.4,np.nan],
        'package':[4,5,6,np.nan,6,7,8,9,np.nan,np.nan]

    }
)

students
```

```python
# sort_values with nan values : 


#--> by default it place all 'nan' values at the end , for placing 'nan' values at top we have to use 'na_position'.

students.sort_values('name')


students.sort_values('name' , na_position = 'first')
```

```python
# sort_values with permanent changes using 'inplace' : 

students.sort_values('name' , na_position = 'first' , ascending=False, inplace = True)
students
```

```python
movies.head()
```

```python
# year wise sorting and then for each year do alphabetical order also : 


movies.sort_values(['year_of_release' , 'title_x'])


movies.sort_values(['year_of_release' , 'title_x'] , ascending = [True , False])  #--> [ year_of_release = True , title_x = False ]
```
