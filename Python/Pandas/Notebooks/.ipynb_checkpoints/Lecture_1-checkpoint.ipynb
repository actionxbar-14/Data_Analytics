```python
# :: Pandas Series : 
# --> A Pandas Series is like a column in a table. It is a 1-D array holding data of any type. 
```

```python
# Importing Pandas : 

import numpy as np
import pandas as pd
```

## Series from lists : 

```python
# String : 

country = [ 'India',  'Pakistan' , 'USA' , 'Nepal' , 'SriLanka']

pd.Series(country)
```

```python
# integers : 

runs = [14 , 24 ,56 , 78 , 100]

pd.Series(runs)
```

```python
# custom index : 

marks = [67 , 57 , 89 , 100]
subjects = ['maths' , 'english' , 'science' , 'hindi']

pd.Series(marks)  #--> index default bnega.

pd.Series(marks , index = subjects)  #--> index as a subject bnega
```

```python
runs = pd.Series(marks)  
```

```python
# setting a name : 

marks = pd.Series(marks , index = subjects , name = 'anubhav ke marks')
marks
```

## Series from dict :

```python
marks = {
    'maths' : 67 ,
    'english' : 57 , 
    'science' : 89 ,
    'hindi' :  100
}

marks_series = pd.Series(marks , name = 'anubhav ke marks')
marks_series
```

### sort_values : 

```python
# values ko sort krr dega ascending/descending order mai , but permanent changes nahi krta hai.
# but agar permanent changes krna hai toh ak parameter hai, [ inplace = True ] krna padta hai. 
```

```python
vk.sort_values()  #--> ascending order
```

```python
vk.sort_values(ascending=False)
```

```python
# Ques : highest runs of vk scored in a single match : 

vk.sort_values(ascending=False).head(1).values[0]
```

```python
# doing permanent changes in series data  :

vk.sort_values(inplace = True)
```

```python
vk
```

### sort_index : 

```python
#  sort_index uss column ko sort krta hai jo table mai as a index use ho rhi h.
#  temporary changes hote hai , permanent changes ke liye [ inplace = True ] krna hota hai.
```

```python
movies.sort_index()
```

```python
# ascending : 

movies.sort_index()
```

```python
# descending : 

movies.sort_index(ascending=False)
```

```python
# permanent_changes : 

movies.sort_index(inplace=True)
```

```python
movies
```
