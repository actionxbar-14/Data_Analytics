```python
# Ques :  why separate objects to handle data and time when python already has datetime functionality? 

# Ans :  

# 1. syntax wise datetime is very convenient.
# 2. But the performance takes a hit while working with huge data. List vs Numpy Array.
# 3. The weaknesses of Python's datetime format inspired the NumPy team to add a set of native time series data type to NumPy.
# 4. The datetime64 dtype encodes dates as 64-bit integers, and thus allows arrays of dates to be represented very compactly.
```

```python
import numpy as np
date = np.array('2015-07-04' , dtype=np.datetime64)
date
```

```python
# addition into current date [ vectorized operation ] :

date + np.arange(12)
```

```python
## NOTE : 

# - Because of the uniform type in NumPy datetime64 arrays, this type of operation can be accomplished much more
#   quickly than if we were working directly with Python's datetime objects, especially as arrays get large. 

# - Pandas Timestamp object combines the ease-of-use of python datetime with 
#   the efficient storage and vectorized interface of numpy.datetime64.

# - From a group of these Timestamp objects, Pandas can construct a DatetimeIndex that can be used to index data in a Series or DataFrame.
```

## DatetimeIndex Object : 

```python
## - A collection of pandas timestamp.
```

```python
## from strings : 

pd.DatetimeIndex( ['2023/1/1' , '2022/1/1' , '2021/1/1'] )[0]
pd.DatetimeIndex( ['2023/1/1' , '2022/1/1' , '2021/1/1'] )[1]
pd.DatetimeIndex( ['2023/1/1' , '2022/1/1' , '2021/1/1'] )[2]
```

```python
## using Python datetime object : 

pd.DatetimeIndex( [dt.datetime(2023,1,1) , dt.datetime(2022,1,1) , dt.datetime(2021,1,1) ] )
```

```python
## using pd.timestamps : 

dt_index = pd.DatetimeIndex( [ pd.Timestamp(2023,1,1) , pd.Timestamp(2022,1,1) , pd.Timestamp(2021,1,1) ] )
dt_index
```

```python
## using datetimeIndex as series index : 



pd.Series( [1,2,3] , index = dt_index)
```

## date_range function : 

```python
##  generally daily dates in a given range :


pd.date_range( start = '2023/1/5' , end = '2023/1/31' , freq = 'D')   #--> Sequentaily date print : D
pd.date_range( start = '2023/1/5' , end = '2023/1/31' , freq = '2D')  #--> alternatively date print : D
```

```python
## business days : [ mon- fri ]

pd.date_range( start = '2023/1/5' , end = '2023/2/28' , freq = 'B')
```

```python
## W -> one week per day : 

pd.date_range( start = '2023/1/5' , end = '2023/2/28') #--> Default : sunday
pd.date_range( start = '2023/1/5' , end = '2023/2/28' , freq = 'W-THU')  #--> Default : Thrusday
```

```python
## h -> Hourly date ( Factor ) :

pd.date_range( start = '2023/1/5' , end = '2023/2/28')  #--> per 1 hour


pd.date_range( start = '2023/1/5' , end = '2023/2/28' , freq = '6h') #--> per 6 hour
```

```python
## ME -> Month end : 

pd.date_range( start = '2023/1/5' , end = '2023/2/28' , freq = 'ME')
```

```python
## M -> Month start : 

pd.date_range( start = '2023/1/5' , end = '2023/2/28' , freq = 'MS')
```

```python
## YE -> Year end : 

pd.date_range( start = '2023/1/5' , end = '2030/1/5/' , freq = 'YE')

```

```python
## using periods( number of results ) :

pd.date_range( start = '2023/1/5' , periods = 25 , freq = 'ME')

```

## to_datetime function : 

```python
#--> converts an existing objects to pandas timestamp/datetimeindex object.
```

```python
# simple series example

s = pd.Series(['2023/1/1','2022/1/1','2021/1/1'])
s.str.split('/').str.get(0)  #--> we use split method to fetch the year because here date is in string format.

#--> Converting it into datetime : 

pd.to_datetime(s).dt.year


pd.to_datetime(s).dt.month  #--> month number
pd.to_datetime(s).dt.month_name()  #--> month name



pd.to_datetime(s).dt.day  #--> day number
pd.to_datetime(s).dt.day_name() #--> day name
```

```python
## with errors : 

s = pd.Series(['2023/1/1','2022/1/1','2021/130/1'])  #--> '2021/130/1' is not valid date
s

pd.to_datetime(s , errors = 'coerce')
```
