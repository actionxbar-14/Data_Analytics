```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

## 1. Relational Plot : 

```python

# - This plot is used to see the statistical relation between 2 or more variables.
# - Used for Bivariate Analysis.

# Plots under this section : 

# - scatterplot
# - lineplot
```

### Scatter plot : 


```python
## It is a axis level function.
## used for the numerical v/s numerical column plots.
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip')
```

```python
# hue parameter : 

sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip' , hue = 'sex')
```

```python
# style parameter : 

sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip' , hue = 'sex' , style = 'time')

```

```python
# size parameter :


sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip' , hue = 'sex' , style = 'time' , size = 'size')
```

```python
## Another way with relplot [ figure level ] functions :
```

```python

sns.relplot(data = tips , x = 'total_bill' , y = 'tip' , kind = 'scatter' , hue = 'sex', style = 'time', size = 'size')
```

### line plot : 

```python
## it is a kind of scatter plot.
## it is a axes level function.
## graph mai agar points ke beech agar continuity dikhani hai toh line plot , agar continuity nhi dikhani toh scatter plot.
```

```python
gap = px.data.gapminder()
gap
```

```python
temp_df = gap[gap['country'] == 'India' ]
temp_df
```

```python
sns.lineplot( data = temp_df , x = 'year' , y = 'lifeExp' , hue = 'country')
```

```python
# hue parameter :

temp_df = gap[gap['country'].isin(['India' , 'Pakistan' , 'China'])]

temp_df
```

```python
## lineplot [axis level ] : 

sns.lineplot(data = temp_df  , x = 'year' , y = 'lifeExp' , hue = 'country')
```

```python
# style parameter : 

```

```python
## Another way with relplot [ figure level ] functions :  
```

```python
sns.relplot(data=temp_df, x='year', y='lifeExp', kind='line')

```

```python

## relplot [figure level ] :

sns.relplot(kind = 'line' , data = temp_df  , x = 'year' , y = 'lifeExp' , hue = 'country')
```
