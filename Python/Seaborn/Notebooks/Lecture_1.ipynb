```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

## 1. Relational Plot : 

```python

# - This plot is used to see the statistical relation between 2 or more variables.
# - Used for Bivariate Analysis.

# Plots under this section : 

# - scatterplot
# - lineplot
# - facet plot
```

### Scatter plot : 


```python
## It is a axis level function.
## used for the numerical v/s numerical column plots.
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip')
```

```python
# hue parameter : 

sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip' , hue = 'sex')
```

```python
# style parameter : 

sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip' , hue = 'sex' , style = 'time')

```

```python
# size parameter :


sns.scatterplot(data = tips , x = 'total_bill' , y = 'tip' , hue = 'sex' , style = 'time' , size = 'size')
```

```python

sns.relplot(data = tips , x = 'total_bill' , y = 'tip' , kind = 'scatter' , hue = 'sex', style = 'time', size = 'size')
```

### line plot : 

```python
## it is a kind of scatter plot.
## it is a axes level function.
## Generally agar x-axis par time hai kisi bhi form mai , month , year , date then we use line plot.
## graph mai agar points ke beech agar continuity dikhani hai toh line plot , agar continuity nhi dikhani toh scatter plot.
```

```python
sns.relplot(data=temp_df, x='year', y='lifeExp', kind='line')

```

```python
## hue parameter : 

temp_df = gap[gap['country'].isin(['India' , 'Pakistan' , 'China'])]

sns.relplot(kind = 'line' , data = temp_df  , x = 'year' , y = 'lifeExp' , hue = 'country')
```

```python
# style parameter : 

temp_df = gap[gap['country'].isin(['India' , 'Brazil' , 'Germany'])]

sns.relplot(kind = 'line' , data = temp_df  , x = 'year' , y = 'lifeExp' , hue = 'country' , style = 'continent')
```

```python
# size parameter : 

temp_df = gap[gap['country'].isin(['India' , 'Brazil' , 'Germany'])]

sns.relplot(kind = 'line' , data = temp_df  , x = 'year' , y = 'lifeExp' , hue = 'country' , style = 'continent' , size = 'continent')
```

### facet plot :

```python
## facet plot ko use krke ham kisi particular categorical column ke upper multiple plot bana skte hoo.
## it will work only with Figure level function Hence -> work with relplot and it will not work with scatterplot and lineplot.
```

```python
tips
```

```python
sns.relplot(data = tips , x = 'total_bill' , y  = 'tip'  ,  kind = 'scatter' , hue = 'sex')

```

```python
#--> facet ( col / row parameter ):

sns.relplot(data = tips , x = 'total_bill' , y  = 'tip'  ,  kind = 'scatter' , row = 'sex')  #--> row
sns.relplot(data = tips , x = 'total_bill' , y  = 'tip'  ,  kind = 'scatter' , col = 'sex')  #--> col
```

```python
# :--> both : 

sns.relplot(data = tips , x = 'total_bill' , y  = 'tip'  ,  kind = 'scatter' , col = 'smoker' , row = 'sex')
```

## 2.  Distribution Plot : 

```python
# - used for univariate analysis.
# - used to find out the distribution.
# - used to find Range of the observation.
# - used to find Central Tendency( mean , median , mode).
# - is the data bimodal?.
# - Are there outliers?.

# Plots under distribution plot :-

# - histplot.
# - kdeplot.
# - rugplot.


# -- > Figure level = displot.
# -- > Axes level  =  histplot , kdeplot , rugplot.
```

### histplot :

```python
## ye data ka distribution batata hai , bins mai convert krke , uss bin mai count krke data ko show krta hai.
## ye generally numerical data ke liye plot kiya jata hai , but categorical column ke liye bhi use hota hai.
```

```python
#--> plotting univariate histogram : 

sns.histplot(data = tips , x = 'total_bill')

```

```python
#--> plotting univariate histogram with displot :

sns.displot(data = tips , x = 'total_bill' , kind = 'hist')
```

```python
## bins parameter : 

sns.displot(data = tips , x = 'total_bill' , kind = 'hist' , bins = 2)
sns.displot(data = tips , x = 'total_bill' , kind = 'hist' , bins = 20)
```

```python
## histplot on categorical column : 


# countplot : 

sns.displot(data = tips , x = 'day' , kind = 'hist')
```

```python
## hue parameter : 
sns.displot(data = tips , x = 'tip' , kind = 'hist' , hue = 'sex')
```

```python
## element -> step parameter : 

sns.displot(data = tips , x = 'tip' , kind = 'hist' , hue = 'sex' , element  = 'step')
```

```python
## titanic dataset :

titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data = titanic , x = 'age' , kind = 'hist'  , element = 'step' , hue = 'sex')
```

```python
## faceting using col and row : 

sns.displot(data = titanic , x = 'age' , kind = 'hist'  , col = 'sex' , element = 'step')
```

```python
# Bivariate histogram : 

# - A bivariate histogram bins the data within rectangles that tile the plot .
# - and then shows the count of observations within each rectangle with the fill color.
```

```python
sns.displot(data = tips , x  = 'total_bill' , y = 'tip' , kind = 'hist')
```

```python
sns.histplot(data = tips , x  = 'total_bill' , y = 'tip' )
```

### kdeplot : 

```python
## kernal density estimation. (kde) 
## Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a continuous density estimate.
```

```python
tips
```

```python
sns.kdeplot(data = tips , x = 'total_bill')
```

```python
## kdeplot using displot : 


sns.displot(data = tips , x = 'total_bill' , kind = 'kde')
```

```python
## hue / fill parameter : 

sns.displot(data = tips , x = 'total_bill' , kind = 'kde' , hue = 'sex' , fill = True)
```

```python
## height , aspect : 

sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,height=10,aspect=2)
```

```python
# Bivariate Kdeplot  : 

# - a bivariate KDE plot smoothes the (x, y) observations with a 2D Gaussian.
```

```python
sns.kdeplot(data = tips , x = 'total_bill' , y = 'tip')
```

### Rugplot : 

```python
# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
```

```python
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

## 3. Matrix Plot : 



```python
# - Heatmap.
# - Clustermap.
```

### Heatmap : 

```python
## It plots a rectangular data as a color-encoded matrix.
## It is a axes level function.
```

```python
gap
```

```python
## converting into wide data  :

temp_df = gap.pivot(index = 'country' , columns = 'year' , values = 'lifeExp')

temp_df
```

```python

plt.figure(figsize = (15,10))


sns.heatmap(temp_df)
```

```python
## annot : [plot with the data given for each matrix box]

temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')


plt.figure(figsize = (15,10))
sns.heatmap(temp_df , annot = True , linewidth = 0.5 , cmap = 'jet')
```
