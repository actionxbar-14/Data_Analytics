# 1. Categorical Scatter Plot : 

## strip plot : 

```python
## ye ak categorical column ko leta hai and ak numerical column ko leta hai and unke beech mai scatter plot ko plot krt hai.
## axis level function.
## data ke beech thoda noise add krke plot krta hai.
```

```python
tips
```

```python
sns.stripplot(data = tips , x = 'day' , y = 'total_bill')
```

```python
## jitter parameter :

sns.stripplot(data = tips , x = 'day' , y = 'total_bill' , jitter = False)
```

## catplot : 

```python
## same functionality as stripplot.
## figure level function.
```

```python
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'strip')
```

```python
## jitter : 

sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'strip' , jitter = 0.2)
```

```python
## hue parameter : 


sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'strip' , jitter = 0.2 , hue = 'sex')
```

## swarmplot : 

```python
## almost equal to the strip plot.
## ye small data ke beech distribution ko easily or ache wayy mai show krta hai.
```

```python
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'swarm')
```

```python
# hue parameter : 

sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'swarm' , hue = 'sex')
```

## Boxplot : 



###

image.png

```python
# A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median,
# third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are.

# Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.
```

```python
sns.boxplot( data = tips , x = 'sex' , y = 'total_bill')
```

```python
sns.boxplot( data = tips , x = 'day' , y = 'total_bill')
```

```python
sns.boxplot( data = tips , x = 'day' , y = 'total_bill' , hue = 'sex')
```

```python
## single boxplot -> numerical column :

sns.boxplot(data = tips , y = 'total_bill')
```

```python
## using catplot : 
```

```python
sns.catplot( data = tips , x = 'day' , y = 'total_bill' , kind = 'box')
```

```python
sns.catplot( data = tips , x = 'day' , y = 'total_bill' , hue = 'sex' , kind = 'box')
```

## Violinplot = (Boxplot + KDEplot) : 

```python
## it is a mixture of both Boxplot and KDE(kernal density estimation) plot.
```

```python
sns.violinplot(data = tips ,x = 'day' , y = 'total_bill')

```

```python
## using catplot : 
sns.catplot(data = tips ,x = 'day' , y = 'total_bill' , kind = 'violin')
sns.catplot(data = tips ,x = 'day' , y = 'total_bill' , kind = 'violin' , hue = 'sex')
```

```python
sns.catplot(data = tips ,x = 'day' , y = 'total_bill' , kind = 'violin' , hue = 'sex' , split = True)
```

# 3. Categorical Estimate Plot : 

```python
## Used for Central tendency.
```

## bar plot : 

```python
## bar plot ak categorical column ka numerical column ke sath ak aggregated value ka graph plot krta hai.
```

```python
sns.barplot(data = tips , x = 'sex' , y = 'total_bill')
```

```python
## estimator parameter : 
```

```python
import numpy as np

sns.barplot(data = tips , x = 'sex' , y = 'total_bill' , hue = 'smoker' , estimator = np.median)
```

sns.barplot(data = tips , x = 'sex' , y = 'total_bill' , hue = 'smoker' , estimator = np.min)

## pointplot  :

```python
## similar to the bar plot.
```

```python
sns.pointplot(data = tips , x = 'sex' , y = 'total_bill' )
```

```python
sns.pointplot(data = tips , x = 'sex' , y = 'total_bill' , hue = 'smoker' )
```

```python

sns.pointplot(data = tips , x = 'sex' , y = 'total_bill' , hue = 'smoker' , estimator = 'max')
sns.pointplot(data = tips , x = 'sex' , y = 'total_bill' , hue = 'smoker' , estimator = 'max' , errorbar  = None)
```

```python
sns.pointplot(data = tips , x = 'sex' , y = 'total_bill' , hue = 'smoker' , estimator = 'min' , errorbar = None)
```

## countplot :

```python
sns.countplot(data = tips , x = 'sex' , hue = 'day')
```

```python
## Faceting : 

sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , col = 'smoker' , kind = 'box' , row = 'time')
```
