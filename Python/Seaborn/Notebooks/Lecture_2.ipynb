# 1. Categorical Scatter Plot : 

## strip plot : 

```python
## ye ak categorical column ko leta hai and ak numerical column ko leta hai and unke beech mai scatter plot ko plot krt hai.
## axis level function.
## data ke beech thoda noise add krke plot krta hai.
```

```python
tips
```

```python
sns.stripplot(data = tips , x = 'day' , y = 'total_bill')
```

```python
## jitter parameter :

sns.stripplot(data = tips , x = 'day' , y = 'total_bill' , jitter = False)
```

## catplot : 

```python
## same functionality as stripplot.
## figure level function.
```

```python
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'strip')
```

```python
## jitter : 

sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'strip' , jitter = 0.2)
```

```python
## hue parameter : 


sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'strip' , jitter = 0.2 , hue = 'sex')
```

## swarmplot : 

```python
## almost equal to the strip plot.
## ye small data ke beech distribution ko easily or ache wayy mai show krta hai.
```

```python
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'swarm')
```

```python
# hue parameter : 

sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'swarm' , hue = 'sex')
```

## Plotting Pairwise Relationship (PairGrid Vs Pairplot) : 

```python
## ye detect krta hai data mai kitne numerical column hai , and detect krne ke baar paris mai graph plot krta hai between numerical columns.
```

```python
iris
```

```python
## pairPlot : 
```

```python
sns.pairplot(iris)
```

```python
sns.pairplot(iris , hue = 'species')
```

```python
## pairGrid : 
```

```python
# pair grid
g = sns.PairGrid(data=iris,hue='species')


# g.map
g.map(sns.scatterplot)
```

```python
# map_diag -> map_upper -> map_lower
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
# vars : 

g = sns.PairGrid(data=iris,hue='species',vars=['sepal_width','petal_width'])


g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```
