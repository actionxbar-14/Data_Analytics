## np.sort

```python
# Returns a sorted of an array.

# documentation : https://numpy.org/doc/stable/reference/generated/numpy.sort.html
```

```python
# syntax : np.sort(array_name , axis , kind , order)
```

 ### Ascending : 




```python
np.sort(a)
```

```python



np.sort(b)  #--> row wise sort

np.sort(b , axis = 0)  #--> column wise sort
```

### descending :



```python
np.sort(a)[::-1]
```

```python



np.sort(b)[::-1]  #--> row wise sort

np.sort(b , axis = 0)[::-1]  #--> column wise sort
```

## np.append : 

```python
# The numpy.append() appends values along the mentioned axis at the end of the array.

# Documentation : https://numpy.org/doc/stable/reference/generated/numpy.append.html
```

```python
# code : 

np.append(a , 200)
```

### Appending an column at the end : 



```python
np.append(b , np.ones((b.shape[0],1)) , axis = 1)
```

### Appending an row at the end : 

```python
np.append(b, np.ones((1, b.shape[1])), axis=0)
```

## np.concatenate : 

```python

# numpy.concatenate() function concatenate a sequence of arrays along an existing axis.

# documentation : https://numpy.org/doc/stable/reference/generated/numpy.concatenate.html
```

```python
# code : 

c = np.arange(6).reshape(2,3)
d = np.arange(6,12).reshape(2,3)

print(c)
print('-' * 50)
print(d)
```

### Row-wise concatenate : 

```python
np.concatenate((c,d) , axis = 1)
```

### Column-wise concatenate : 


```python
np.concatenate((c,d) , axis = 0)
```

## np.expand_dims : 

```python

# With the help of Numpy.expand_dims() method, we can get the expanded dimensions of an array. [1D ko 2D and 2D ko 3D mai convert karna]. 

# Documentation :  https://numpy.org/doc/stable/reference/generated/numpy.expand_dims.html
```

```python
a.shape
```

### 1D array ko 2D main convert karna : [ row vector ]



```python

np.expand_dims(a , axis = 0)
```

```python
np.expand_dims(a , axis = 0).shape
```

### 1D array ko 2D main convert karna : [ column vector ]



```python

np.expand_dims(a , axis = 1)
```

## np.where : 

```python

# The numpy.where() function returns the indices of elements in an input array where the given condition is satisfied.

# Documentation : https://numpy.org/doc/stable/reference/generated/numpy.where.html
```

```python
# syntax :  np.where(condition , true , false) [ agar condition true hai toh true wala chlega , otherwise false wala chlega ].
```

```python
a
```

```python
# find all indices with value greater than 50 : 

np.where(a > 50 )
```

```python
# replace all values > 50 with 0 :

np.where(a > 50 , 0, a)
```

## np.argmax / np.argmin : 




```python

# The numpy.argmax()/numpy.argmin function returns indices of the max/min element of the array in a particular axis.

# Documentatiton : https://numpy.org/doc/stable/reference/generated/numpy.argmax.html
```

### np.argmax

```python
a
```

```python
np.argmax(a)  #--> max element ka index return krega.
```

```python
b 
```

```python
np.argmax(b , axis = 1)  #--> return indexes of max element in the row.
```

```python
b
```

```python
np.argmin(b , axis = 1)  #--> return indexes of min element in the row.
```

## np.cumsum : 

```python

# numpy.cumsum() function is used when we want to compute the cumulative sum of array elements over a given axis.

# Documentation : https://numpy.org/doc/stable/reference/generated/numpy.cumsum.html
```

```python
a
```

```python
np.cumsum(a)
```

```python
b
```

```python
# without axis :

np.cumsum(b)
```

```python
# With axis :  column

np.cumsum(b , axis = 0)

```

```python
# With axis : row

np.cumsum(b , axis = 1)
```
