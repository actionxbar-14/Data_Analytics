# Creating Numpy Arrays :

```python

# np.array :   [ 1D array ]   --> 'Vector' 

import numpy as np


a = np.array([1,2,3])
print(a)
print(type(a))
```

```python
# np.array :   [ 2D array ]   --> 'Matrix'


import numpy as np


b = np.array([ [1,2,3], [4,5,6] ])
print(b)
print(type(b))
```

```python
# np.array :   [ 3D array ]   --> 'Tensor'


import numpy as np


c = np.array([[[1,2], [3,4]],[[5,6] , [7,8]]])

print(c)
print(type(c))
```

## dtype : 

```python

# 1D array with float value : 

np.array([1,2,3] , dtype = float)
```

```python
# 1D array with bool value :
np.array([1,2,3] , dtype = bool)
```

## np.arange :   

```python
# it gives us tha facility to form the array with a given range. (same as range in python )
# syntax : np.arange(start , stop , step)
```

```python
np.arange(1 , 11)
```

```python
np.arange(1 ,11 ,2)
```

```python
# with reshape :

np.arange(1 , 11).reshape(2 ,5)
```

## np.ones and np.zeroes : 

```python
# it provide us the matrix or array containing all ones or all zeroes as elements. 
# it is generally used to initialize the value of an array. 
# it also used to generate a array with random numbers. 
```

```python
np.ones((3 ,4))
```

```python
np.zeros((3,4))
```

## np.random : 

```python
# random numbers within 0 to 1 :

np.random.random((3 , 4))
```

## np.linspace :

```python
# ye linearly seperable points generate krta hai in a given range. 
# ml algorithm ka result dekhne ke kaam aata h. 
```

```python
np.linspace(-10 , 10 , 10)
```

## np.identity :


```python
# provide the identity matrix.
```

```python
np.identity(4)
```
