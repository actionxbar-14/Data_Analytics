# Creating Numpy Arrays :

```python

# np.array :   [ 1D array ]   --> 'Vector' 

import numpy as np


a = np.array([1,2,3])
print(a)
print(type(a))
```

```python
# np.array :   [ 2D array ]   --> 'Matrix'


import numpy as np


b = np.array([ [1,2,3], [4,5,6] ])
print(b)
print(type(b))
```

```python
# np.array :   [ 3D array ]   --> 'Tensor'


import numpy as np


c = np.array([[[1,2], [3,4]],[[5,6] , [7,8]]])

print(c)
print(type(c))
```

## dtype :

# Array Functions : 


```python
# creating array : 

a1 = np.random.random((3,3))
a1 = np.round(a1 * 100 )
```

```python
a1
```

## max/min/sum/prod :

```python
# max :

np.max(a1)


```

```python
# min :

np.min(a1)
```

```python
# sum :

np.sum(a1)
```

```python
# prod : 

np.prod(a1)
```

```python
# har row/column ka minimun/maximum value  : 

# axis =  0 --> column ,
# axis =  1 --> row.



```

```python
np.max(a1 , axis = 0)  #--> column wise max value .
```

```python
np.max(a1 , axis = 1)  #--> row wise max value .
```

## mean / meadian / std / var  :


```python
# mean : 

np.mean(a1 , axis = 1)
np.mean(a1 , axis = 0)
```

```python
# median : 

np.median(a1 , axis = 1)
np.median(a1 , axis = 0)
```

```python
# standard deviation :

np.std(a1 , axis = 1)
np.std(a1 , axis = 0)
```

```python
# variance : 

np.var(a1 , axis = 1)
np.var(a1 , axis = 0)
```

## Trigonometric Functions :

```python
np.sin(a1)
```

## dot product : 

```python
# pahle wale matrix ka column  , dusre wale matrix ke row se match karna chaiye. 
```

```python
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(12 , 24).reshape(4,3)

np.dot(a2,a3)
```

```python
np.exp(a1)
```
