# Creating Numpy Arrays :

```python

# np.array :   [ 1D array ]   --> 'Vector' 

import numpy as np


a = np.array([1,2,3])
print(a)
print(type(a))
```

```python
# np.array :   [ 2D array ]   --> 'Matrix'


import numpy as np


b = np.array([ [1,2,3], [4,5,6] ])
print(b)
print(type(b))
```

```python
# np.array :   [ 3D array ]   --> 'Tensor'


import numpy as np


c = np.array([[[1,2], [3,4]],[[5,6] , [7,8]]])

print(c)
print(type(c))
```

# Stacking : 

## horizontal stacking :


```python
a4 = np.arange(12).reshape(3,4)

a5 = np.arange(12 , 24).reshape(3,4)
```

```python
a4

```

```python
a5
```

```python
# horizontal stacking : 

np.hstack((a4 , a5))
```

## Vertical Stacking : 

```python
a6 = np.arange(12).reshape(3,4)

a7 = np.arange(12 , 24).reshape(3,4)
```

```python
np.vstack((a6 , a7))
```
