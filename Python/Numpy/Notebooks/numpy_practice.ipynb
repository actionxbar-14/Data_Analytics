```python
## array attribute : 

# np.arange :-

a = np.arange(12)    #--> [ 1D array - [vector] ]
print(a)
print(type(a))
print(a.ndim)
print(a.shape)
print(a.size)
print(a.itemsize)
print(a.dtype)
print(a.astype(np.int32))

print('-' * 50 )

a = np.arange(16).reshape(4,4)    #--> [ 2D array - [matrix] ]
print(a)
print(type(a))
print(a.ndim)
print(a.shape)
print(a.size)
print(a.itemsize)
print(a.dtype)
print(a.astype(np.int32))

print('-' * 50 )

a = np.arange(18).reshape(3,2,3)    #--> [ 3D array - [tensor] ]
print(a)
print(type(a))
print(a.ndim)
print(a.shape)
print(a.size)
print(a.itemsize)
print(a.dtype)
print(a.astype(np.int32))

```
