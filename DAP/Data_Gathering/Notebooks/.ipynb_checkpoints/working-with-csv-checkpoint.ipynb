## 2. Opening a csv file from an URL :


```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

## 13. Convertors : 


```python
# --> Convertors ak aisa method hai jiska main use hota hai data ke kisi column ko explicitely 
#     kisi function ka use krke transform krna.

# --> Ex : 
# Ipl dataset mai team1 name ke column mai team ka full name likha hai like ( Royal challengers Bengalore ) ,
# par hame chaiye sirf iska short form ( RCB ) toh iske liye converters use hoga. 
```

```python
# pd.read_csv(r"C:\Users\ANUBHAV\Data_Analytics\DAP\Data_Gathering\Datasets\IPL Matches 2008-2020.csv")

# Explicit function : 

def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB" 
    else:
        return name 
```

```python
rename("Royal Challengers Bangalore")
```

```python
# Implimentation : 

pd.read_csv(r"C:\Users\ANUBHAV\Data_Analytics\DAP\Data_Gathering\Datasets\IPL Matches 2008-2020.csv" , converters = {'team1': rename})
        
```
