```python
# 127 pages ke data ko dataframe mai add krne ke liye loop chalaya hai or har bar data ko temp_df variable mai store krke usse actual
```

```python
for i in range(1,128):
    response = requests.get(f'https://api.themoviedb.org/3/tv/top_rated?api_key=2907a3004f657179e96619acd03cc0f8&page={i}')
    temp_df =  pd.DataFrame(response.json()['results'])[['id' , 'name' , 'overview' , 'first_air_date' , 'popularity' , 'vote_average' , 'vote_count']]                       
    df = pd.concat([df, temp_df], ignore_index=True)
    
```
