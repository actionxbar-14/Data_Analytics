```python
import pandas as pd
df = pd.DataFrame() 
```

# 1. to_csv : 

```python
df = pd.read_csv(r"C:\Users\ANUBHAV\Data_Analytics\DAP\Data_Gathering\Datasets\deliveries.csv")
df.head() # --> starting ki 5 rows ko print kreaga. 

```

```python
df.groupby('batsman')['batsman_runs'].sum()  #--> data ko batsman and batsman ke runs ke according club krega. 
```

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index() #--> uss data ko temp_df mai store krega
```

```python
temp_df  
```

```python
temp_df.to_csv('batsman_runs.csv')  #--> temp mai jo created data hai usse as a csv file store krr dega current directory mai. 
```

```python
# creating pivot table : 


df.pivot_table(index = 'batsman' , columns = 'bowling_team' , aggfunc = 'sum' )


```

```python
# storing the pivot_table : 


df.pivot_table(index = 'batsman' , columns = 'bowling_team' , aggfunc = 'sum' ).to_csv('batsman_vs_team.csv' , index = False)
```

## 2. to_excel : 

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index()  # --> uss data ko temp_df mai store krega
```

```python
temp_df
```

```python
temp_df.to_excel("batsman_runs.xlsx")   #--> temp mai jo created data hai usse as a excel file store krr dega current directory mai. 
```

```python
# storing the data in excel with particular sheet:  

# :--> sheet_1 :
temp_df.to_excel("output_runs.xlsx" , sheet_name = 'batsman_runs' )
```

```python
# :--> sheet_2 :
temp_df2 = df.pivot_table(index = 'batsman' , columns = 'bowling_team' , aggfunc = 'sum' )
```

```python
with pd.ExcelWriter('Output.xlsx') as writer:                               #--> ExcelWriter ak excel file mai do alag - alag sheet banata hai           
    temp_df.to_excel(writer , sheet_name = 'sheet_name_1')
    temp_df2.to_excel(writer , sheet_name = 'sheet_name_2')
       
```

## 3. to_html : 
#  blog form mai html table mai data mo store krta hai . 

```python
df.pivot_table(index = 'over' , columns = 'ball' , values = 'batsman_runs' , aggfunc = 'count') 
```

```python
df.query('batsman_runs == 6').pivot_table(index = 'over' , columns = 'ball' , values = 'batsman_runs' , aggfunc = 'count') # ye query pivot table
# mai se kisi bhi over mai , kisi bhi ball pe historically kitne six lagte hai. 
```

```python
df.query('batsman_runs == 6').pivot_table(index = 'over' , columns = 'ball' , values = 'batsman_runs' , aggfunc = 'count').to_html('sixes_heatmap.html')
```
