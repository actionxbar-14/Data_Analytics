```python
# 10. Concatenate and Merge Dataframe ( JOINS ) :
```

```python
df1 = pd.DataFrame( { 'ID' : [1,2,3] , 'Name' : ['A' , 'B' , 'C'] } )
df1
```

```python
df2 = pd.DataFrame( { 'ID' : [1,2,2,4] , 'Score' : [88 , 96 , 77 , 79] } )
df2
```

```python
# concatenate :
```

```python
pd.concat([df1 , df2] , axis = 0)  #-->vertical / row level / top on top. 
```

```python
pd.concat([df1 , df2] , axis = 0)  #--> horizontal / column level / side by side. 
```

```python
# merge / Join :
```

```python
pd.merge(df1 , df2 , how = 'inner' , on = 'ID')
```

```python
# join when common column have different name :
```

```python
pd.merge(df1 , df2 , how = 'inner' , left_on = 'ID' , right_on = 'ID')
```
