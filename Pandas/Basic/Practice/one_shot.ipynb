```python
data =  {
    'Name' : ['Anubhav' , 'Div' , 'Sakshi' , 'Aryan'] ,
    'age'  : [21 ,24 ,22 , 23],
    'Gender' : ['male' , 'male' , 'Female' , 'male']   
}
```

```python
#:--> df.rename() : pahli baat toh ye hai ki iska use hota hai column name ko change krne ke liye , dusri baat ye hai ki ye original DataFrame
# ko change nhi krta hai , hame uss nayi changed column vali table ko ak naye variable mai store krke rkhna hota hai.

df1 = df.rename( columns = {'age' : 'Age'})
print(df)

print(" " * 50)
print(df1)
```

```python
# NOTE: Existing DataFrame mai change krne ke liye hame ak argument ka use krna hota hai jo default mai False rhti hai , inplace = False , isko
# inplace = True krne se original DataFrame mai Column name change ho jayegi. 

print(df)

df.rename( columns = { 'age' : 'Age'} , inplace = True)

print(" " * 50)

print(df)
```
